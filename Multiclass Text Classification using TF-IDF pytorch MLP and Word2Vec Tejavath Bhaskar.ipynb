## 9. Data Cleaning and Text Preprocessing

> Text Preprocessing
> "Text preprocessing is traditionally an important step for Natural Language Processing (NLP) tasks. It transforms text into a more digestible form so that machine learning algorithms can perform better."
> *Source: GeeksforGeeks - Text Preprocessing in Python.*

Raw text may contain:

- Uppercase/lowercase variations
- URLs
- HTML tags
- Punctuation
- Extra spaces
- Missing values

We will perform basic text cleaning using regular expressions to normalize the text.

Important:

We should not blindly remove every type of information from text.

The preprocessing strategy should depend on the dataset and the model.

````python
def clean_text(text):

    # Convert to string
    text = str(text)

    # Convert to lowercase
    text = text.lower()

    # Remove URLs
    text = re.sub(
        r"http\S+|www\S+|https\S+",
        "",
        text
    )

    # Remove HTML tags
    text = re.sub(
        r"<.*?>",
        "",
        text
    )

    # Remove punctuation
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    # Remove extra spaces
    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text
````

````python
train_df["clean_Description"] = (
    train_df["Description"]
    .apply(clean_text)
)
test_df["clean_Description"] = (
    test_df["Description"]
    .apply(clean_text)
)
````

## 12. Label Encoding

Machine learning algorithms—including Logistic Regression and Neural Networks—are mathematical models. They operate exclusively on **numerical data** (vectors and matrices) and cannot directly process raw text labels like "sports" or "business."

### Why do we need Label Encoding?

Before we can feed our target variable (`Category`) into a model, we must convert the human-readable string labels into a machine-readable format. **Label Encoding** is the standard approach for this; it assigns a unique integer to each distinct category in the dataset.

### The Transformation Process

When we use the `LabelEncoder` from Scikit-learn, the process follows these steps:

1.  **Scanning:** The encoder scans the entire label column in the training set to identify all unique categories.
2.  **Mapping:** It assigns a unique integer to each category (typically in alphabetical order).
3.  **Transformation:** It replaces every string label in the original data with its corresponding assigned integer.

**Mapping Example:**

| Original Label (String) | Encoded Label (Integer) |
| :--- | :--- |
| `business` | 0 |
| `sci/tech` | 1 |
| `sports` | 2 |
| `world` | 3 |

### Implementation Guidelines

*   **Fitting vs. Transforming:**
    *   Use `.fit_transform()` only on the **training data** so the encoder learns the unique categories and defines the mapping.
    *   Use only `.transform()` on the **validation and test data**. This ensures the test set uses the *exact same mapping* learned from the training set. If you "fit" the encoder on test data, the model might assign different integers to the same classes, leading to incorrect evaluations.
*   **Reversing the Encoding:** Since models output predictions as integers (e.g., `2`), you can use `label_encoder.inverse_transform([2])` at any time to convert the prediction back into the human-readable string (`"sports"`).

````python
from sklearn.preprocessing import LabelEncoder

# Initialize the encoder
label_encoder = LabelEncoder()

# Fit and transform the training labels
y_train_encoded = label_encoder.fit_transform(y_train)

# Transform the validation and test labels using the learned mapping
y_val_encoded = label_encoder.transform(y_val)
y_test_encoded = label_encoder.transform(test_df["Category"])

# Display the mapping
print("Classes learned by the encoder:")
for index, label in enumerate(label_encoder.classes_):
    print(f"{index} → {label}")
````

## 13. Feature Extraction: TF-IDF

> **Definition: TF-IDF (Term Frequency-Inverse Document Frequency)**
> "TF-IDF is a statistical measure that evaluates how relevant a word is to a document in a collection of documents. This is done by multiplying two metrics: how many times a word appears in a document, and the inverse document frequency of the word across a set of documents."
> *Source: Introduction to Information Retrieval by Christopher D. Manning, Prabhakar Raghavan, and Hinrich Schütze (Cambridge University Press).*

**Mathematical representation of the TF-IDF calculation**

The TF-IDF score for a word ($w$) in a document is calculated using these three components:

1. **Term Frequency (TF):** Measures how often a word appears in a document.
   $$TF(w) = \frac{\text{Number of times word } w \text{ appears in the document}}{\text{Total number of words in the document}}$$

2. **Inverse Document Frequency (IDF):** Measures the significance of a word across the entire corpus. Rare words get higher scores.
   $$IDF(w) = \log \left( \frac{\text{Total number of documents}}{\text{Number of documents containing the word } w} + 1 \right)$$

3. **TF-IDF Score:** The final weight is the product of the two.
   $$TF\text{-}IDF(w) = TF(w) \times IDF(w)$$

*Source: GeeksforGeeks - NLP Representations.*

Machine learning models cannot directly work with raw text.

We therefore convert text into numerical representations.

There are several ways to represent text:

For our first experiment, we will use **TF-IDF**.

## 19. Cross Entropy Loss

> **Cross-Entropy**
> "Cross-entropy loss, or log loss, measures the performance of a classification model where the prediction input is a probability value between 0 and 1. The loss increases as the predicted probability diverges from the actual label."
> *Source: Deep Learning by Ian Goodfellow (MIT Press).*

In PyTorch, the standard loss function for multi-class classification is:

    nn.CrossEntropyLoss()

To understand how it works, we need to look at the underlying mathematics. PyTorch's `CrossEntropyLoss` actually combines two mathematical operations into a single, numerically stable step: **Softmax** and **Negative Log-Likelihood**.

### 1. The Softmax Function (Logits to Probabilities)
Our neural network outputs raw, unnormalized scores called **logits** ($z$). To interpret these as probabilities ($p$), we apply the Softmax function. For a specific class $i$ out of $C$ total classes:

$$p_i = \frac{e^{z_i}}{\sum_{j=1}^{C} e^{z_j}}$$

This ensures that all predicted probabilities fall between $0$ and $1$, and sum up to exactly $1.0$.

### 2. Negative Log-Likelihood (Calculating the Error)
Once we have the probabilities, we calculate the error using the Cross-Entropy formula. For a single text sample, the loss $L$ is defined as:

$$L = -\sum_{i=1}^{C} y_i \log(p_i)$$

Where $y_i$ represents the true label (a one-hot encoded vector where the correct class is $1$ and all others are $0$). Because $y_i$ is $0$ for all incorrect classes, the formula simplifies to just looking at the probability assigned to the **correct class** ($c$):

$$L = -\log(p_c)$$

### Conceptual Example (Our 4 Categories)

Suppose our classes are `Business`, `Sci/Tech`, `Sports`, and `World`.

If the true class is **Sports** ($y = [0, 0, 1, 0]$), the loss is simply $-\log(p_{\text{Sports}})$.

*   **Confident and Correct:** If the model predicts $p_{\text{Sports}} = 0.90$, then $L = -\log(0.90) \approx 0.105$. (The loss is small).
*   **Confident but Wrong:** If the model predicts $p_{\text{Sports}} = 0.05$ (meaning it highly predicted another class), then $L = -\log(0.05) \approx 2.99$. (The loss is much larger).

Cross-Entropy strongly penalizes models that are highly confident about the wrong answer.

---

## Important PyTorch Detail: Logits vs. Probabilities

Because computing the exponent in the Softmax function can cause mathematical overflows (numbers getting too large for the computer's memory), PyTorch calculates the logarithm and the Softmax together.

Mathematically, PyTorch simplifies the combined formula to:

$$L = -z_c + \log\left(\sum_{j=1}^{C} e^{z_j}\right)$$

Where $z_c$ is the raw logit score of the correct class.

**What this means for your code:**
Our model must output **raw logits** (e.g., `[0.5, -1.2, 2.5, 0.3]`), not probabilities. We pass these logits directly to `nn.CrossEntropyLoss()`.

You should **NOT** apply a Softmax layer manually at the end of your `nn.Module` before calculating the loss.

print(f"Calculated Loss: {loss.item():.4f}")

The basic PyTorch training process is:

3. Perform forward pass
4. Calculate loss

6. Update parameters

The key code sequence is:

          ↓
    outputs = model(...)
          ↓

## 25. Multi-Layer Perceptron (MLP)

> **Definition: Multi-Layer Perceptron**
> "A multilayer perceptron (MLP) is a class of feedforward artificial neural network (ANN)... An MLP consists of at least three layers of nodes: an input layer, a hidden layer and an output layer."
> *Source: Neural Networks and Deep Learning by Michael Nielsen.*

**[Diagram: Architecture of a deep neural network]**
<img src="https://cs231n.github.io/assets/nn1/neural_net2.jpeg" alt="MLP Architecture" width="500">

Logistic Regression contains one linear transformation:

    Input
      ↓
    Linear
      ↓
    Output

An MLP introduces one or more hidden layers and nonlinear activation functions.

For example:

    Input
      ↓
    Linear
      ↓
    ReLU
      ↓
    Linear
      ↓
    Output

The nonlinear activation allows the network to learn nonlinear relationships.

Without nonlinear activation functions, stacking multiple linear layers
still results in an overall linear transformation.

### Flexible MLP Model

````python
class MLP(nn.Module):

    def __init__(
        self,
        input_dim,
        hidden_dims,
        num_classes,
        activation
    ):

        super().__init__()

        layers = []

        previous_dim = input_dim

        for hidden_dim in hidden_dims:

            layers.append(
                nn.Linear(
                    previous_dim,
                    hidden_dim
                )
            )

            layers.append(
                activation()
            )

            previous_dim = hidden_dim

        layers.append(
            nn.Linear(
                previous_dim,
                num_classes
            )
        )

        self.network = nn.Sequential(
            *layers
        )

    def forward(self, x):

        return self.network(x)
````

The function will:

1. Train using mini-batches
2. Calculate training loss
3. Calculate validation loss
4. Store loss values
5. Return training history

This allows us to reuse exactly the same training procedure for:

- ReLU
- Sigmoid
- Tanh
- Leaky ReLU
- GELU
- Different MLP depths
- Different optimizers

````python
def train_model(
    model,
    train_loader,
    val_loader,
    optimizer,
    criterion,
    epochs=10,
    device=device
):

    train_losses = []
    val_losses = []

    for epoch in range(epochs):

        # ==================================
        # Training
        # ==================================

        model.train()

        running_train_loss = 0.0

        for X_batch, y_batch in train_loader:

            X_batch = X_batch.to(device)
            y_batch = y_batch.to(device)

            optimizer.zero_grad()

            outputs = model(X_batch)

            loss = criterion(
                outputs,
                y_batch
            )

            loss.backward()

            optimizer.step()

            running_train_loss += (
                loss.item() *
                X_batch.size(0)
            )

        epoch_train_loss = (
            running_train_loss /
            len(train_loader.dataset)
        )

        # ==================================
        # Validation
        # ==================================

        model.eval()

        running_val_loss = 0.0

        with torch.no_grad():

            for X_batch, y_batch in val_loader:

                X_batch = X_batch.to(device)
                y_batch = y_batch.to(device)

                outputs = model(X_batch)

                loss = criterion(
                    outputs,
                    y_batch
                )

                running_val_loss += (
                    loss.item() *
                    X_batch.size(0)
                )

        epoch_val_loss = (
            running_val_loss /
            len(val_loader.dataset)
        )

        train_losses.append(
            epoch_train_loss
        )

        val_losses.append(
            epoch_val_loss
        )

        print(
            f"Epoch [{epoch+1}/{epochs}] "
            f"Train Loss: {epoch_train_loss:.4f} "
            f"Val Loss: {epoch_val_loss:.4f}"
        )

    return train_losses, val_losses
````

### Prediction Function

````python
def predict_model(
    model,
    data_loader,
    device=device
):

    model.eval()

    all_predictions = []

    with torch.no_grad():

        for X_batch, _ in data_loader:

            X_batch = X_batch.to(device)

            outputs = model(X_batch)

            predictions = torch.argmax(
                outputs,
                dim=1
            )

            all_predictions.extend(
                predictions.cpu().numpy()
            )

    return np.array(all_predictions)
````

## 33. Loss vs. Accuracy Explained

Loss and accuracy are related, but they are NOT the same.

## Accuracy

Accuracy asks:

    How many predictions were correct?

For example:

    90 correct out of 100

gives:

    Accuracy = 90%

---

## Loss

Loss considers how confident the model was.

Suppose the correct class is Sports.

Prediction A:

    Sports      = 0.90
    Politics    = 0.05
    Technology  = 0.05

Prediction B:

    Sports      = 0.51
    Politics    = 0.48
    Technology  = 0.01

Both may be classified as Sports.

Therefore, both may contribute equally to accuracy.

However, their confidence levels are different.

Cross Entropy Loss captures this difference.

---

## During training

The optimizer minimizes the **loss**.

It does not directly optimize accuracy.

The training process is:

    Prediction
        ↓
    Loss
        ↓
    Gradient
        ↓
    Optimizer
        ↓
    Updated Parameters

````python
import numpy as np

def sentence_to_vector(tokens, wv, vector_size):
    """Average the Word2Vec vectors of in-vocabulary tokens.
    Return a zero vector of shape (vector_size,) if there are none.
    """
    vectors = [wv[token] for token in tokens if token in wv]

    if not vectors:
        return np.zeros(vector_size, dtype=np.float32)

    return np.mean(vectors, axis=0).astype(np.float32)


# Build sentence-level Word2Vec features
X_train_w2v = np.array(
    [sentence_to_vector(tokens, word2vec_model, EMBED_DIM)
     for tokens in train_tokens],
    dtype=np.float32
)

X_val_w2v = np.array(
    [sentence_to_vector(tokens, word2vec_model, EMBED_DIM)
     for tokens in val_tokens],
    dtype=np.float32
)

X_test_w2v = np.array(
    [sentence_to_vector(tokens, word2vec_model, EMBED_DIM)
     for tokens in test_tokens],
    dtype=np.float32
)

print("X_train_w2v shape:", X_train_w2v.shape)
print("X_val_w2v shape:", X_val_w2v.shape)
print("X_test_w2v shape:", X_test_w2v.shape)
````

### Check: does `model_config` match your saved weights?

This loads your saved `.pt` file and compares the `model_config` you declared against the actual shapes of the weights in `model_state_dict` — a common mistake is changing `hidden_dims` after training without updating the checkpoint-saving cell to match. If they disagree, the grader can't reconstruct your model, so this check catches it now and prints exactly what was declared vs. what was actually saved.

````python
import torch

def check_model_config_matches_weights(pt_path):
    ckpt = torch.load(pt_path, map_location="cpu", weights_only=True)
    mc = ckpt["model_config"]
    sd = ckpt["model_state_dict"]

    linear_weight_keys = sorted(
        [k for k in sd if k.endswith(".weight") and sd[k].dim() == 2],
        key=lambda k: int(k.split(".")[1]),
    )
    shapes = [sd[k].shape for k in linear_weight_keys]
    actual_input_dim = shapes[0][1]
    actual_num_classes = shapes[-1][0]
    actual_hidden_dims = [s[0] for s in shapes[:-1]]

    declared_input_dim = mc["input_dim"]
    declared_hidden_dims = list(mc["hidden_dims"])
    declared_num_classes = mc["num_classes"]

    ok = (
        declared_input_dim == actual_input_dim
        and declared_hidden_dims == actual_hidden_dims
        and declared_num_classes == actual_num_classes
    )

    if ok:
        print(f"✅ model_config matches the saved weights.")
        print(f"   input_dim={actual_input_dim}, hidden_dims={actual_hidden_dims}, num_classes={actual_num_classes}")
        return True

    print(f"❌ model_config does NOT match the saved weights.\n")
    print(f"{'':12s}{'declared (model_config)':<30s}{'actual (model_state_dict)':<30s}")
    print(f"{'input_dim':12s}{str(declared_input_dim):<30s}{str(actual_input_dim):<30s}"
          f"{'  <-- MISMATCH' if declared_input_dim != actual_input_dim else ''}")
    print(f"{'hidden_dims':12s}{str(declared_hidden_dims):<30s}{str(actual_hidden_dims):<30s}"
          f"{'  <-- MISMATCH' if declared_hidden_dims != actual_hidden_dims else ''}")
    print(f"{'num_classes':12s}{str(declared_num_classes):<30s}{str(actual_num_classes):<30s}"
          f"{'  <-- MISMATCH' if declared_num_classes != actual_num_classes else ''}")
    print(
        f"\nThis usually means hidden_dims (or the architecture) changed after "
        f"mlp_w2v was trained, but the checkpoint-saving cell's model_config "
        f"wasn't updated to match. Fix model_config in the saving cell so it "
        f"exactly matches what you passed to MLP(...) when you trained mlp_w2v, "
        f"then re-run the saving cell and this check again."
    )
    return False


assert check_model_config_matches_weights(f"{STUDENT_ID}_word2vec_mlp.pt"), (
    "model_config does not match the saved weights -- see the verbose output above."
)
````
